```python
# Importing libraries
import os
import cv2
from PIL import Image
import sys # Import sys for sys.exit()

# Set path to the Google Drive folder with images
from google.colab import drive
drive.mount('/content/drive')
path = "/content/drive/MyDrive/ResNetModel/outputs/prediction_results"

# Check if the directory exists before attempting to change to it
if not os.path.exists(path):
    print(f"Error: The specified path '{path}' does not exist. Please ensure the folder structure is correct in your Google Drive and try again.")
    sys.exit("Exiting script due to invalid path.") # Exit if the path is invalid

try:
    # Attempt to change directory. If this fails, the script exits.
    os.chdir(path)
except OSError as e:
    print(f"Error changing directory to '{path}': {e}. Please check permissions or if the directory is accessible.")
    sys.exit("Exiting script due to directory access error.") # Exit if chdir fails

mean_height = 0
mean_width = 0

# --- Start of modified section for error handling and path consistency ---
try:
    # Get list of image files directly from 'path'
    image_files = [file for file in os.listdir(path) if file.lower().endswith( (".jpg", ".jpeg", ".png") )]
    num_of_images = len(image_files)
    print("Number of Images:", num_of_images)

    if num_of_images == 0:
        print(f"Error: No image files found in the directory '{path}'. Please ensure images are present and have '.jpg', '.jpeg', or '.png' extensions.")
        sys.exit("Exiting script as no images were found.")

    # Calculating the mean width and height of all images
    for file in image_files: # Iterate over the filtered list
        im = Image.open(os.path.join(path, file))
        width, height = im.size
        mean_width += width
        mean_height += height

    # Averaging width and height
    mean_width = int(mean_width / num_of_images)
    mean_height = int(mean_height / num_of_images)

    # Resizing all images to the mean width and height
    for file in image_files: # Iterate over the filtered list
        im = Image.open(os.path.join(path, file))
        # Use Image.LANCZOS instead of Image.ANTIALIAS for downsampling
        im_resized = im.resize((mean_width, mean_height), Image.LANCZOS)
        # Save back to the original file path within the specified directory
        im_resized.save(os.path.join(path, file), 'JPEG', quality=95)
        print(f"{file} is resized")

    # Function to generate video
    def generate_video():
        image_folder = path
        video_name = 'PupilWithPredictions.avi'

        # Use the already filtered image_files list
        images = image_files
        print("Images:", images)

        if not images:
            print("Error: No images found to create video. Exiting video generation.")
            return

        # Set frame from the first image
        first_image_path = os.path.join(image_folder, images[0])
        frame = cv2.imread(first_image_path)
        if frame is None:
            print(f"Error: Could not read the first image '{first_image_path}'. It might be corrupted or inaccessible.")
            return

        height, width, layers = frame.shape

        # Video writer to create .avi file. Video will be saved in 'path'
        video_output_path = os.path.join(image_folder, video_name)
        video = cv2.VideoWriter(video_output_path, cv2.VideoWriter_fourcc(*'DIVX'), 1, (width, height))

        # Appending images to video
        for image_name in images:
            img_full_path = os.path.join(image_folder, image_name)
            img = cv2.imread(img_full_path)
            if img is not None:
                video.write(img)
            else:
                print(f"Warning: Could not read image '{image_name}', skipping.")

        # Release the video file
        video.release()
        cv2.destroyAllWindows()
        print(f"Video generated successfully at: {video_output_path}")

    # Calling the function to generate the video
    generate_video()

except OSError as e:
    print(f"\nCritical I/O error during file operations in '{path}': {e}.")
    print("This typically indicates a problem with accessing the Google Drive folder. Please check the following:")
    print("1. **Google Drive Mount Status**: Ensure your Google Drive is still correctly mounted and accessible. Sometimes mounts can become unstable.")
    print("2. **Path Accuracy**: Double-check that the path '/content/drive/MyDrive/ResNetModel/outputs/prediction_results' exists and is spelled correctly within your Google Drive.")
    print("3. **Folder Permissions**: Verify that the specified Google Drive folder has sufficient read/write permissions for your Colab environment.")
    print("You might need to remount Google Drive or restart the Colab runtime if the issue persists.")
    sys.exit("Exiting script due to critical file system access error.")
# --- End of modified section ---
```
